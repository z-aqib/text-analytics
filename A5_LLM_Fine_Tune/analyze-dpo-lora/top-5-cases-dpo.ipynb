```python
import pandas as pd
```

```python
# Load the CSV from the 'deliverables' folder
file_path = 'deliverables/ITA A5 - DPO.csv'
df = pd.read_csv(file_path)
```

```python
df
```

```python
# Filter for specific case numbers
case_numbers = [1, 4, 5, 6, 7]
filtered_df = df[df['CASE GIT'].isin(case_numbers)]
```

```python
# Drop the 'name' column if it exists
if 'NAME' in filtered_df.columns:
    filtered_df = filtered_df.drop(columns=['NAME'])
```

```python
# Rename the column
if 'CASE GIT' in filtered_df.columns:
    filtered_df = filtered_df.rename(columns={'CASE GIT': 'CASE'})
    
    # Rename the column
if 'EXPLANATION' in filtered_df.columns:
    filtered_df = filtered_df.rename(columns={'EXPLANATION': 'CHANGE'})
```

```python
# Save the filtered dataframe to a new CSV file
output_path = 'deliverables/dpo_top5_experiments.csv'
filtered_df.to_csv(output_path, index=False)

print(f"Filtered data saved to '{output_path}'")
```

# remove name from all experiments
