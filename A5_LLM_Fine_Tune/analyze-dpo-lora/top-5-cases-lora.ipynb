```python
import pandas as pd
```

```python
# Load the CSV from the 'deliverables' folder
file_path = 'deliverables/ITA A5 - LORA.csv'
df = pd.read_csv(file_path)
```

```python
df
```

```python
# Filter for specific case numbers
case_numbers = [1, 7, 10, 20, 26]
filtered_df = df[df['CASE ACCORDING TO ZUHA GITHUB'].isin(case_numbers)]
```

```python
# Drop the 'name' column if it exists
if 'NAME' in filtered_df.columns:
    filtered_df = filtered_df.drop(columns=['NAME'])
```

```python
# Rename the column
if 'CASE ACCORDING TO ZUHA GITHUB' in filtered_df.columns:
    filtered_df = filtered_df.rename(columns={'CASE ACCORDING TO ZUHA GITHUB': 'CASE'})
```

```python
# Save the filtered dataframe to a new CSV file
output_path = 'deliverables/lora_top5_experiments.csv'
filtered_df.to_csv(output_path, index=False)

print(f"Filtered data saved to '{output_path}'")
```
